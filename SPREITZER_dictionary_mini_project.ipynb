```python
#stats from chatgpt, I asked it to use spotify
#the first keys are the artist names, then I have subdictionaries that follow this format:
'''
monthlyListenersMillions = number (in millions) of unique people that have listened to the artist in the past 28 days as of 10/21/26
peakMonthlyListnersMillions = all time high of the above
totalStreamsBillions = total number of streams that artist has on spotify as of 10/21/26
totalTracks = how many songs/tracks the artist has on spotify as of 10/21/26
topSongs lists the 3 most streamed songs of the artist and their number of streams in billions as of 10/21/26

'''
musicArtists = {
    'Bruno Mars': {
        'monthlyListenersMillions': 128.7,
        'peakMonthlyListenersMillions': 151.1,
        'totalStreamsBillions': 59.7,
        'totalTracks': 124,
        'topSongs': {
            'Die With A Smile': {
                'streamsBillions': 4.08
            },
            'Locked out of Heaven': {
                'streamsBillions': 3.56
            },
            'Just the Way You Are': {
                'streamsBillions': 3.53
            }
        }
    },
    'Rihanna': {
        'monthlyListenersMillions': 116.8,
        'peakMonthlyListenersMillions': 117.0,
        'totalStreamsBillions': 63.5,
        'totalTracks': 386,
        'topSongs': {
            'We Found Love': {
                'streamsBillions': 2.71
            },
            'Umbrella': {
                'streamsBillions': 2.45
            },
            'This Is What You Came For': {
                'streamsBillions': 2.42
            }
        }
    },
    'The Weeknd': {
        'monthlyListenersMillions': 116.7,
        'peakMonthlyListenersMillions': 126.2,
        'totalStreamsBillions': 99.3,
        'totalTracks': 320,
        'topSongs': {
            'Blinding Lights': {
                'streamsBillions': 5.64
            },
            'Starboy': {
                'streamsBillions': 4.82
            },
            'Die For You': {
                'streamsBillions': 3.47
            }
        }
    },
    'Justin Bieber': {
        'monthlyListenersMillions': 111.7,
        'peakMonthlyListenersMillions': 147.0,
        'totalStreamsBillions': 81.2,
        'totalTracks': 357,
        'topSongs': {
            'STAY': {
                'streamsBillions': 4.11
            },
            'Love Yourself': {
                'streamsBillions': 3.40
            },
            'Let Me Love You': {
                'streamsBillions': 3.27
            }
        }
    },
    'Taylor Swift': {
        'monthlyListenersMillions': 105.8,
        'peakMonthlyListenersMillions': 116.2,
        'totalStreamsBillions': 132.4,
        'totalTracks': 670,
        'topSongs': {
            'Cruel Summer': {
                'streamsBillions': 3.52
            },
            'Blank Space': {
                'streamsBillions': 2.64
            },
            'I Don’t Wanna Live Forever': {
                'streamsBillions': 2.22
            }
        }
    }
}

#SPECIAL FEATURES:
#avg. streams per song
#percentage of total streams top 3 make up
#rank artists on individual stats

'''
OPTION 1: FINISHED (print dict.)
OPTION 2: FINISHED (add artist)
OPTION 3: FINISHED (remove artist)
OPTION 4: FINISHED (avg. streams per song)
OPTION 5: FINISHED (top 3 songs % of total streams)
OPTION 6: UNFINISHED (ranking stats between artists) (not started)
OPTION 7: FINISHED (exit)

'''
def artist_input():  #This function makes my code a little smoother, so instead of having to write the following line 16 different times, I just call the function when I need an artist inputted from the user
  user_artist = input("What artist would you like to choose?\n").title
  return user_artist


def add_artist(user_artist):
  musicArtists[user_artist] = {
        'monthlyListenersMillions': float(input("Monthly listeners in millions:")),
        'peakMonthlyListenersMillions': float(input("Monthly listeners all time high in millions:")),
        'totalStreamsBillions': float(input("Total streams in billions:")),
        'totalTracks': int(input("Total tracks:")),
        'topSongs': {
            input("Top 1 song name:").title: {"streams": float(input("Number of streams: in billions"))},
            input("Top 2 song name:").title: {"streams": float(input("Number of streams: in billions"))},
            input("Top 3 song name:").title: {"streams": float(input("Number of streams: in billions:"))}
        }
        }
  print(f"{user_artist} has been added to the dictionary.")


def rank_stat(user_stat):
  ranked_artists = sorted(
        musicArtists,
        key = lambda artist: musicArtists[artist][user_stat],
        reverse = True
    )
  for i, artist in enumerate(ranked_artists):
    print(f"{i+1}. {artist}")


def top_3_percent(user_artist, total_streams_top3):
  for song in musicArtists[user_artist]["topSongs"]:
    total_streams_top3 += musicArtists[user_artist]["topSongs"][song]["streamsBillions"]
  percent = (total_streams_top3 / (musicArtists[user_artist]["totalStreamsBillions"])) * 100
  print(f"{user_artist}'s top 3 songs consist of {percent}% of their total streams on spotify.")




while True:
  while True:
    print("Choose an option number from the following:")
    print("1. Print dictionary")
    print("2. Add item to dictionary")
    print("3. Remove item from dictionary")
    print("4. Print average number of streams per song")
    print("5. Print top 3 songs' percentage of total streams")
    print("6. Rank a certain stat between artists")
    print("7. Exit")

    try:
      choose = int(input("Choose: "))
      if choose >= 1 and choose <= 7:
        break
      else:
        print("Please enter a number from 1 to 7.\n")
    except ValueError:
      print("Invalid input. Please enter a whole number.\n")

  if choose == 1:  #PRINT DICTIONARY
  #I don't think this requires an explanation
    print(musicArtists)
    for artist, info in musicArtists.items():
      print(f'\n{artist}: ')
      for stat, value in info.items():
        if stat == "topSongs":
          print(f'  {stat.title()}:')
          for song, song_info in value.items():
            print(f'    {song}')
            for song_stat, song_value in song_info.items():
              print(f'      {song_stat.title()}: {song_value}')
        else:
          print(f'  {stat}: {value}')


  elif choose == 2:  #ADD ARTIST
    #This option asks for a bunch of inputs to fill in a new artist's information. It follows the format I talked about at the beginning.
    #I used the template we got in class to format my inputs.
    user_artist = artist_input()
    add_artist(user_artist)


  elif choose == 3:  #REMOVE ARTIST
    #Pretty self-explanatory, user inputs an artist to remove and my code removes it from the dictionary. The artist could also be one they added to the dictionary themselves.
    user_artist = artist_input()
    musicArtists.pop(user_artist)
    print(f"{user_artist} has been removed.")

  elif choose == 4:  #AVG STREAMS PER SONG
  #I used Direct Access to find the elements I needed from the dictionary, then divided them within the print line.
    user_artist = artist_input()
    print(f"{user_artist}'s average number of streams per song is {(musicArtists[user_artist]["totalStreamsBillions"])/(musicArtists[user_artist]["totalTracks"])*1000} million.\n\n")

  elif choose == 5:  #TOP 3 SONGS % OF TOTAL
    user_artist = artist_input()
    total_streams_top3 = 0  #This is a placeholder value
    top_3_percent(user_artist, total_streams_top3)


  elif choose == 6:  #RANKING STATS
    user_stat = input(f"What stat would you like to rank?")
    #The following bit I googled a lot about how to do this and this is basically what is happening:
    #1. my dictionary is being sorted (.sorted()). The artist names are going into a new list in alphabetical order.
    #2. the lambda part essentially does this:
    '''
    def get_stat(artist):
    return musicArtists[artist][user_stat]
    '''
    #Then, the reversed makes sure that the highest values start, then decrease. (example: 1. [3] 2. [2] 3. [1])
    #Then, my for loop loops through the list I just made and prints the list in the format 1. [artist] 2. [artist] et cetera
    #(Because the index starts at 0, I have to add the "+1" or else the ranked list will print as 0. through 4.)
    rank_stat(user_stat)


  elif choose == 7:  #EXIT
    #This option breaks the while True loop
    break

  else:
    #sort of error handling, I'll finish this later (CHANGE TO TRY/EXCEPT)
    print("That is not a valid input. Please enter a number 1-7.\n")
```

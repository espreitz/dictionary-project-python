```python
#stats from chatgpt, I asked it to use spotify
#the first keys are the artist names, then I have subdictionaries that follow this format:
'''
Monthly Listeners Millions = number (in millions) of unique people that have listened to the artist in the past 28 days as of 10/21/26
Peak Monthly Listners Millions = all time high of the above
Total Streams Billions = total number of streams that artist has on spotify as of 10/21/26
Total Tracks = how many songs/tracks the artist has on spotify as of 10/21/26
Top Songs lists the 3 most streamed songs of the artist and their number of streams in billions as of 10/21/26
'''

musicArtists = {
    'Bruno Mars': {
        'Monthly Listeners Millions': 128.7,
        'Peak Monthly Listners Millions': 151.1,
        'Total Streams Billions': 59.7,
        'Total Tracks': 124,
        'Top Songs': {
            'Die With A Smile': {
                'Streams Billions': 4.08
            },
            'Locked out of Heaven': {
                'Streams Billions': 3.56
            },
            'Just the Way You Are': {
                'Streams Billions': 3.53
            }
        }
    },
    'Rihanna': {
        'Monthly Listeners Millions': 116.8,
        'Peak Monthly Listners Millions': 117.0,
        'Total Streams Billions': 63.5,
        'Total Tracks': 386,
        'Top Songs': {
            'We Found Love': {
                'Streams Billions': 2.71
            },
            'Umbrella': {
                'Streams Billions': 2.45
            },
            'This Is What You Came For': {
                'Streams Billions': 2.42
            }
        }
    },
    'The Weeknd': {
        'Monthly Listeners Millions': 116.7,
        'Peak Monthly Listners Millions': 126.2,
        'Total Streams Billions': 99.3,
        'Total Tracks': 320,
        'Top Songs': {
            'Blinding Lights': {
                'Streams Billions': 5.64
            },
            'Starboy': {
                'Streams Billions': 4.82
            },
            'Die For You': {
                'Streams Billions': 3.47
            }
        }
    },
    'Justin Bieber': {
        'Monthly Listeners Millions': 111.7,
        'Peak Monthly Listners Millions': 147.0,
        'Total Streams Billions': 81.2,
        'Total Tracks': 357,
        'Top Songs': {
            'STAY': {
                'Streams Billions': 4.11
            },
            'Love Yourself': {
                'Streams Billions': 3.40
            },
            'Let Me Love You': {
                'Streams Billions': 3.27
            }
        }
    },
    'Taylor Swift': {
        'Monthly Listeners Millions': 105.8,
        'Peak Monthly Listners Millions': 116.2,
        'Total Streams Billions': 132.4,
        'Total Tracks': 670,
        'Top Songs': {
            'Cruel Summer': {
                'Streams Billions': 3.52
            },
            'Blank Space': {
                'Streams Billions': 2.64
            },
            'I Don’t Wanna Live Forever': {
                'Streams Billions': 2.22
            }
        }
    }
}

#SPECIAL FEATURES:
#avg. streams per song
#percentage of total streams top 3 make up
#rank artists on individual stats

def artist_input():  #This function makes my code a little smoother, so instead of having to write the following line 16 different times, I just call the function when I need an artist inputted from the user
  while True:
    user_artist = input("What artist would you like to choose?\n").title()
    if user_artist in musicArtists:
      return user_artist
    else:
      print("That artist is not in the dictionary. Please enter a valid input.")

def add_artist(user_artist):
  musicArtists[user_artist] = {
        'Monthly Listeners Millions': float(input("Monthly listeners in millions:")),
        'Peak Monthly Listners Millions': float(input("Monthly listeners all time high in millions:")),
        'Total Streams Billions': float(input("Total streams in billions:")),
        'Total Tracks': int(input("Total tracks:")),
        'Top Songs': {
            input("Top 1 song name:").title(): {"Streams Billions": float(input("Number of streams: in billions"))},
            input("Top 2 song name:").title(): {"Streams Billions": float(input("Number of streams: in billions"))},
            input("Top 3 song name:").title(): {"Streams Billions": float(input("Number of streams: in billions:"))}
        }
        }
  print(f"{user_artist} has been added to the dictionary.")

def rank_stat(user_stat):
  ranked_artists = sorted(
        musicArtists,
        key = lambda artist: musicArtists[artist][user_stat],
        reverse = True
    )
  for i, artist in enumerate(ranked_artists):
    print(f"{i+1}. {artist}")

def top_3_percent(user_artist, total_streams_top3):
  for song in musicArtists[user_artist]["Top Songs"]:
    total_streams_top3 += musicArtists[user_artist]["Top Songs"][song]["Streams Billions"]
  percent = (total_streams_top3 / (musicArtists[user_artist]["Total Streams Billions"])) * 100
  print(f"{user_artist}'s top 3 songs consist of {percent}% of their total streams on spotify.")

while True:
  while True:
    print("Choose an option number from the following:")
    print("1. Print dictionary")
    print("2. Add item to dictionary")
    print("3. Remove item from dictionary")
    print("4. Print average number of streams per song")
    print("5. Print top 3 songs' percentage of total streams")
    print("6. Rank a certain stat between artists")
    print("7. Exit")

    try:
      choose = int(input("Choose: "))
      if choose >= 1 and choose <= 7:
        break
      else:
        print("Please enter a number from 1 to 7.\n")
    except ValueError:
      print("Invalid input. Please enter a whole number.\n")

  if choose == 1:  #PRINT DICTIONARY
  #I don't think this requires an explanation
    print(musicArtists)
    for artist, info in musicArtists.items():
      print(f'\n{artist}: ')
      for stat, value in info.items():
        if stat == "Top Songs":
          print(f'  {stat.title()}:')
          for song, song_info in value.items():
            print(f'    {song}')
            for song_stat, song_value in song_info.items():
              print(f'      {song_stat.title()}: {song_value}')
        else:
          print(f'  {stat}: {value}')


  elif choose == 2:  #ADD ARTIST
    #This option asks for a bunch of inputs to fill in a new artist's information. It follows the format I talked about at the beginning.
    #I used the template we got in class to format my inputs.
    user_artist = artist_input()
    add_artist(user_artist)


  elif choose == 3:  #REMOVE ARTIST
    #Pretty self-explanatory, user inputs an artist to remove and my code removes it from the dictionary. The artist could also be one they added to the dictionary themselves.
    user_artist = artist_input()
    musicArtists.pop(user_artist)
    print(f"{user_artist} has been removed.")

  elif choose == 4:  #AVG STREAMS PER SONG
  #I used Direct Access to find the elements I needed from the dictionary, then divided them within the print line.
    user_artist = artist_input()
    print(f"{user_artist}'s average number of streams per song is {(musicArtists[user_artist]["Total Streams Billions"])/(musicArtists[user_artist]["Total Tracks"])*1000} million.\n\n")

  elif choose == 5:  #TOP 3 SONGS % OF TOTAL
    user_artist = artist_input()
    total_streams_top3 = 0  #This is a placeholder value
    top_3_percent(user_artist, total_streams_top3)

  elif choose == 6:  #RANKING STATS
    while True:
      existing_artist = input("Please enter an artist that is currently in the dictionary (this helps the code run correctly).\n").title()
      if existing_artist in musicArtists:
        break
      else:
        print("That artist isn't in the dictionary. Please enter an artist that is in the dictionary.")
    #For the function, I googled a lot about how to do this and this is basically what is happening:
    #1. my dictionary is being sorted (.sorted()). The artist names are going into a new list in alphabetical order.
    #2. the lambda part essentially does this:
    '''
    def get_stat(artist):
    return musicArtists[artist][user_stat]
    '''
    #Then, the reversed makes sure that the highest values start, then decrease. (example: 1. [3] 2. [2] 3. [1])
    #Then, my for loop loops through the list I just made and prints the list in the format 1. [artist] 2. [artist] et cetera
    #(Because the index starts at 0, I have to add the "+1" or else the ranked list will print as 0. through 4.)
    while True:
      user_stat = input(f"What stat would you like to rank?\n").title()
      if user_stat in musicArtists[existing_artist]:
        if user_stat != "Top Songs":
          break
        else:
          print("You can't rank top songs of artists (yet).")
      else:
        print("That statistic is not in the dictionary. Please enter a valid input.")

    rank_stat(user_stat)

  elif choose == 7:  #EXIT
    #This option breaks the while True loop
    break
```
